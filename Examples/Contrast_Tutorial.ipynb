## Contrast Tests for Functional ANOVA  
### Zero-Sum Contrast Requirement

In functional ANOVA, contrast tests evaluate linear combinations of group mean functions. A contrast is defined by a weight vector

$$
\mathbf{c} = (c_1, c_2, \dots, c_G)
$$

and specifies the hypothesis

$$
H_0:\; \sum_{g=1}^{G} c_g \mu_g(t) = 0 
\quad \text{for all } t.
$$

---

### Zero-Sum Constraint

For a valid **contrast comparison between groups**, the weights must satisfy

$$
\sum_{g=1}^{G} c_g = 0.
$$

This ensures that:

- The hypothesis represents a **pure difference between group means**.
- The test is invariant to an overall shift in the functional level.
- The contrast compares groups relative to each other rather than to zero.

---

### Interpretation

When  

$$
\sum_{g=1}^{G} c_g = 0,
$$

the hypothesis

$$
\sum_{g=1}^{G} c_g \mu_g(t) = 0
$$

tests whether a specific linear combination of **group differences** equals zero across the domain.

For example:

- $$\mathbf{c} = [-1, 1]$$  
  tests  
  $$
  -\mu_1(t) + \mu_2(t) = 0,
  $$
  i.e., whether the two mean functions are equal.

- $$\mathbf{c} = [0, -1, 0, 1]$$  
  tests  
  $$
  -\mu_2(t) + \mu_4(t) = 0,
  $$
  comparing two selected levels.

- $$\mathbf{c} = [-3, 3, -2, 2]$$  
  satisfies  
  $$
  -3 + 3 - 2 + 2 = 0,
  $$
  and tests a weighted comparison between the specified groups.

---

### Scaling Property

If $\mathbf{c}$ is a valid contrast (i.e., sums to zero), then any nonzero scalar multiple $a\mathbf{c}$ defines the **same hypothesis**:

$$
\sum_{g=1}^{G} a c_g \mu_g(t) = 0.
$$

Thus, scaling changes only the magnitude of the test statistic, not the null hypothesis.

---

All functional ANOVA inference procedures (L2, F-type, simultaneous tests, and Behrens–Fisher variants) apply to valid zero-sum contrasts.

```python
from scipy.io import loadmat
import os
import numpy as np
import pandas as pd
from functionalANOVA.fanova import functionalANOVA
```

```python
# Import Data
matlab_data = loadmat("../Data/example_data.mat")

# Get data out of .mat file
groups = [matlab_data['TwoWayData'][0, 0], matlab_data['TwoWayData'][0, 1]]
time = matlab_data['timeData']
indicator_list = [matlab_data['IndicatorCell'][0,0], matlab_data['IndicatorCell'][0,1]]
```

```python
# Bounds on time
bounds = (-np.inf, np.inf)
n_boot = 1000 # Default it 10,000 (small for tutorial examples)
seed = 1
myANOVA = functionalANOVA(data_list=groups, d_grid=time, grid_bounds=bounds, subgroup_indicator=indicator_list,
                          n_boot=n_boot,
                          primary_labels=["Control", "Treatment"],
                          secondary_labels = ["Slow", "Normal", "Fast", "Very Fast"],
                          domain_label='Time',
                          domain_units='seconds',
                          response_label="Measurement Value",
                          response_units="a.u.")
```

## Custom Contrast Tests (Zero-Sum Required)

All custom contrasts must satisfy

$$
\sum_{g=1}^{G} c_g = 0.
$$

This ensures the hypothesis represents a pure comparison between group means.

---

### 1️⃣ Treatment vs Control

Contrast:
$$
\mathbf{c} = [-1,\, 1]
$$

Check:
$$
-1 + 1 = 0
$$

Hypothesis:
$$
-\mu_1(t) + \mu_2(t) = 0
$$

This tests whether the two primary groups have equal mean functions.

```python
myANOVA.twoway(hypothesis="custom", methods = ["L2-Simul", 'F-Simul'],  contrast = [-1, 1], seed=seed)
```

### 2️⃣ Normal vs Very Fast

Contrast:
$$
\mathbf{c} = [0,\, -1,\, 0,\, 1]
$$

Check:
$$
0 - 1 + 0 + 1 = 0
$$

Hypothesis:
$$
-\mu_2(t) + \mu_4(t) = 0
$$

This compares the "Normal" and "Very Fast" secondary levels.

```python
myANOVA.twoway(hypothesis="custom",  contrast = [0, -1, 0, 1], seed=seed)
```

### 3️⃣ Weighted Multi-Group Comparison

Contrast:
$$
\mathbf{c} = [-3,\, 3,\, -2,\, 2]
$$

Check:
$$
-3 + 3 - 2 + 2 = 0
$$

Hypothesis:
$$
-3\mu_1(t) + 3\mu_2(t) - 2\mu_3(t) + 2\mu_4(t) = 0
$$

This tests a weighted comparison across multiple groups.

```python
myANOVA.twoway(hypothesis="custom",  contrast = [-3, 3, -2, 2], seed=seed)
```
